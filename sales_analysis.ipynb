```python
region_state_city = df.groupby(['region','state','city'])['sales'].sum().round(2).sort_values(ascending=False).reset_index()
```

```python
result = []
for i in region_state_city['region'].unique():


  region_data = region_state_city[region_state_city['region'] == i]

  state_sales = region_data.groupby('state')['sales'].sum().round(2).sort_values(ascending=False)

  top_state = state_sales.index[0]

  state_data = region_data[region_data['state'] == top_state]



  city_sales = state_data.groupby('city')['sales'].sum().round(2).sort_values(ascending=False)

  top_city = city_sales.index[0]



  result.append([i,top_state,top_city,city_sales[top_city]])



result = pd.DataFrame(result,columns=['region','top_state','top_city','sales'])

print(result)




```

```python
result['location'] = result['region'] + ' -> ' + result['top_state'] + ' -> ' + result['top_city']
```

```python
plt.barh(
    result['location'],
    result['sales']
)

plt.xlabel('Total Sales ($)')
plt.ylabel('Location')
plt.title('Top Selling City In Each Region')

plt.show()
```
